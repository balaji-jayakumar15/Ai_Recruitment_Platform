```python
features = [
    "Job_Role",
    "Location",
    "Experience_Level",
    "Experience_Years",
    "Education",
    "Technical_Skills",
    "Skill_Match_Percent",
    "Preferred_Skill_Match_Percent",
    "Experience_Score",
    "Project_Match_Percent",
    "Education_Score",
    "ATS_Score"
]

target = "Avg_Salary_LPA"

X=df[features]
y=df[target]

print("Feature",X.shape)
print("Target",y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training",X_train.shape)
print("Testing",X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "Job_Role",
    "Location",
    "Experience_Level",
    "Education",
    "Technical_Skills"
]

numerical_features = [
    "Experience_Years",
    "Skill_Match_Percent",
    "Preferred_Skill_Match_Percent",
    "Experience_Score",
    "Project_Match_Percent",
    "Education_Score",
    "ATS_Score"
]

preprocessor=ColumnTransformer(
    transformers=[
        ("categorical",
         OneHotEncoder(handle_unknown="ignore"),
         categorical_features),
         ("numerical",
          "passthrough",
          numerical_features)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

salary_model=Pipeline(
    steps=[
        ("Preprocessing",preprocessor),
        ("Regressor",
         RandomForestRegressor(
             n_estimators=200,
             random_state=42,
             n_jobs=-1
         ))
    ]
)
```

```python
salary_model.fit(X_train,y_train)
print("Model trained successfully")
```

```python
y_predict=salary_model.predict(X_test)

print("First 10 predictions:")

for actual, predicted in zip(y_test.head(10), y_predict[:10]):
    print(
        f"Actual: {actual:.2f} LPA | "
        f"Predicted: {predicted:.2f} LPA"
    )
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

mae=mean_absolute_error(y_test,y_predict)

rmse=np.sqrt(
    mean_squared_error(y_test,y_predict)
)

r2= r2_score(y_test,y_predict)

print("="*50)
print("Model-5 Salary Prediction")
print("="*50)
print("MAE",mae)
print("RMSE",rmse)
print("R^2", r2)
```

```python
import joblib

joblib.dump(salary_model,"Salary_Prediction.pkl")

print("Model-5 Saved Successfully")
```
